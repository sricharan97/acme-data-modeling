```python
%%sql
-- Drop existing tables to start fresh
DROP TABLE IF EXISTS user_ratings CASCADE;
DROP TABLE IF EXISTS purchases CASCADE;
DROP TABLE IF EXISTS customer_addresses CASCADE;
DROP TABLE IF EXISTS customer_emails CASCADE;
DROP TABLE IF EXISTS customer_phones CASCADE;
DROP TABLE IF EXISTS products CASCADE;
DROP TABLE IF EXISTS customers CASCADE;

-- TODO: write your queries here
-- Core Customer table
CREATE TABLE Customers(
  customer_id INT PRIMARY KEY,
  first_name VARCHAR(100) NOT NULL,
  last_name VARCHAR(100) NOT NULL,
  created_at TIMESTAMP NOT NULL
);

-- Core Products table
CREATE TABLE Products(

    product_id INT PRIMARY KEY,
    product_name VARCHAR(150) NOT NULL,
    product_type VARCHAR(150) NOT NULL,
    description TEXT,
    price DECIMAL(10,2) NOT NULL CHECK(price >= 0.00),
    stock_quantity INT NOT NULL CHECK(stock_quantity >= 0),
    created_at TIMESTAMP NOT NULL
);

-- Customer Contact Channels (one - many relationships)
CREATE TABLE Customer_Phones(
    
    phone_id INT PRIMARY KEY,
    customer_id INT NOT NULL REFERENCES Customers(customer_id) on DELETE CASCADE,
    phone VARCHAR(50) NOT NULL,
    created_at TIMESTAMP NOT NULL,
    CONSTRAINT uq_customer_phone UNIQUE (customer_id , phone)

);

CREATE TABLE Customer_Emails(
    
    email_id INT PRIMARY KEY,
    customer_id INT NOT NULL REFERENCES Customers(customer_id) on DELETE CASCADE,
    email VARCHAR(150) NOT NULL,
    created_at TIMESTAMP NOT NULL,
    CONSTRAINT uq_customer_email UNIQUE (customer_id , email)

);

CREATE TABLE Customer_Addresses (
    
    address_id INT PRIMARY KEY,
    customer_id INT NOT NULL REFERENCES Customers(customer_id) ON DELETE CASCADE,
    address VARCHAR(255) NOT NULL,
    created_at TIMESTAMP NOT NULL,
    CONSTRAINT uq_customer_address UNIQUE (customer_id, address)
);

-- 4. REFACTORED: Purchases Table (3NF-compliant Transaction Ledger)
-- Replaces rigid 'gear' columns with a direct Product Foreign Key reference.
-- The total_price column is intentionally stored as an immutable transaction audit 
-- snapshot. While it equals quantity * unit_price at execution time, storing it explicitly
-- preserves point-in-time financial records against future catalog price adjustments or 
-- historical overrides.
CREATE TABLE Purchases(
    purchase_id INT,
    customer_id INT NOT NULL REFERENCES Customers(customer_id) ON DELETE RESTRICT,
    product_id INT NOT NULL REFERENCES Products(product_id) ON DELETE RESTRICT,
    quantity INT NOT NULL CHECK (quantity > 0),
    unit_price DECIMAL(10, 2) NOT NULL CHECK (unit_price >= 0.00),
    total_price DECIMAL(10, 2) NOT NULL CHECK (total_price >= 0.00),
    purchase_date TIMESTAMP NOT NULL,
    status VARCHAR(50) NOT NULL,
    PRIMARY KEY (purchase_id, product_id) -- 3NF Composite Line-Item Key
);

-- Ratings table (many - many feedback junction)

CREATE TABLE User_Ratings (
    
    rating_id INT PRIMARY KEY,
    customer_id INT NOT NULL REFERENCES Customers(customer_id) ON DELETE CASCADE,
    product_id INT NOT NULL REFERENCES Products(product_id) ON DELETE CASCADE,
    rating INT NOT NULL CHECK (rating >= 1 AND rating <= 5),
    review_text TEXT,
    rating_date TIMESTAMP NOT NULL,
    CONSTRAINT uq_customer_product_rating UNIQUE (customer_id, product_id)
);


```

## Loading Test Data

```python
from sqlalchemy import create_engine, inspect as sa_inspect, text

engine = create_engine('postgresql://postgres:postgres@127.0.0.1:5432/postgres')

# Comment out any tables you haven't created yet to build up your schema gradually
tables = [
    "customers",
    "customer_phones",
    "customer_emails",
    "customer_addresses",
    "products",
    "purchases",
    "user_ratings",
]

# Verify all tables in the list exist before attempting to load
existing = sa_inspect(engine).get_table_names()
missing = [t for t in tables if t not in existing]
if missing:
    raise RuntimeError(f"Tables not found: {missing} — run your CREATE TABLE DDL first.")

# Print column types and primary/foreign keys for each table
print("Schema overview:")
with engine.connect() as conn:
    for table in tables:
        print(f"\n  {table}")
        cols = conn.execute(text(
            "SELECT column_name, data_type FROM information_schema.columns "
            "WHERE table_schema = 'public' AND table_name = :t ORDER BY ordinal_position"
        ), {"t": table})
        for col in cols:
            print(f"    {col.column_name}: {col.data_type}")
        keys = conn.execute(text(
            "SELECT tc.constraint_type, kcu.column_name "
            "FROM information_schema.table_constraints tc "
            "JOIN information_schema.key_column_usage kcu "
            "  ON tc.constraint_name = kcu.constraint_name "
            "  AND tc.table_schema = kcu.table_schema "
            "WHERE tc.table_schema = 'public' AND tc.table_name = :t "
            "AND tc.constraint_type IN ('PRIMARY KEY', 'FOREIGN KEY') "
            "ORDER BY tc.constraint_type, kcu.column_name"
        ), {"t": table})
        for key in keys:
            print(f"    [{key.constraint_type}] {key.column_name}")

# Load data with 3NF Transformations
print("\nLoading data...")
for table in tables:
    csv_file_path = os.path.join(csv_path, f'{table}.csv')
    
    # -------------------------------------------------------------------------
    # Scenario A: Standard Tables (Direct Load)
    # -------------------------------------------------------------------------
    if table not in ["products", "purchases"]:
        df = pd.read_csv(csv_file_path)
        df.to_sql(table, engine, if_exists='append', index=False)
        print(f"  {table}.csv → {table} table (Directly loaded)")
        
    # -------------------------------------------------------------------------
    # Scenario B: Products Ingestion (Ensuring 3NF Catalog Items exist)
    # -------------------------------------------------------------------------
    elif table == "products":
        # Ensure our gear definitions are explicitly present in the catalog
        # to satisfy foreign key checks in the purchases ledger
        gear_products = pd.DataFrame([
            {
                "product_id": 104,
                "product_name": "Large Gear",
                "product_type": "Industrial Component",
                "description": "Heavy-duty 3D printed mechanical gear",
                "price": 50.00,
                "stock_quantity": 100,
                "created_at": pd.Timestamp.now()
            },
            {
                "product_id": 105,
                "product_name": "Small Gear",
                "product_type": "Precision Component",
                "description": "High-precision 3D printed micro gear",
                "price": 25.00,
                "stock_quantity": 250,
                "created_at": pd.Timestamp.now()
            }
        ])
        
        # Load custom products if they exist in a CSV, otherwise load gears
        if os.path.exists(csv_file_path):
            df = pd.read_csv(csv_file_path)
            # Combine the CSV catalog with the gear definition dataframe, removing duplicates
            df = pd.concat([df, gear_products]).drop_duplicates(subset=['product_id'])
        else:
            df = gear_products
            
        df.to_sql("products", engine, if_exists='append', index=False)
        print("  Gears & Catalog Products → products table (3NF catalog established)")
        
         # -------------------------------------------------------------------------
    # Scenario C: Purchases Transformation (Unpivoting Flat Gears to Line Items)
    # -------------------------------------------------------------------------
    elif table == "purchases":
        df_flat = pd.read_csv(csv_file_path)
        normalized_records = []
        
        # Unpivot each flat row into individual line items
        for _, row in df_flat.iterrows():
            # Check for Large Gear purchase
            if row.get("large_gear_quantity", 0) > 0:
                qty = int(row["large_gear_quantity"])
                unit_price = float(row.get("large_gear_unit_price", 50.00))
                normalized_records.append({
                    "purchase_id": row["purchase_id"],
                    "customer_id": row["customer_id"],
                    "product_id": 104,  # Large Gear ID
                    "quantity": qty,
                    "unit_price": unit_price,
                    "total_price": qty * unit_price,
                    "purchase_date": row["purchase_date"],
                    "status": row["status"]
                })
                
            # Check for Small Gear purchase
            if row.get("small_gear_quantity", 0) > 0:
                qty = int(row["small_gear_quantity"])
                unit_price = float(row.get("small_gear_unit_price", 25.00))
                normalized_records.append({
                    "purchase_id": row["purchase_id"],
                    "customer_id": row["customer_id"],
                    "product_id": 105,  # Small Gear ID
                    "quantity": qty,
                    "unit_price": unit_price,
                    "total_price": qty * unit_price,
                    "purchase_date": row["purchase_date"],
                    "status": row["status"]
                })
        
        df_normalized = pd.DataFrame(normalized_records)
        df_normalized.to_sql("purchases", engine, if_exists='append', index=False)
        print(f"  {table}.csv → transformed into {len(df_normalized)} 3NF line items → purchases table")

print("\n" + "="*50)
print("DATABASE INITIALIZATION & 3NF DATA LOAD: SUCCESSFUL")
print("="*50)

```

## NoSQL Modeling Design Decisions

1. **Why are recent purchases and industry products embedded in the customer document?**

Recent purchases and industry recommendations are embedded directly into the customer document to optimize for read heavy query patterns on key user facing pages (such as the account dashboard, order history preview).

2. **What are the trade-offs of embedding vs. referencing?**

Embedding (Denormalization):

Pros: Optimized read performance (single lookup returns all required data), eliminates complex $lookup aggregation joins, and aligns well with document-oriented data locality.

Cons: Potential data duplication, risk of data staleness (e.g., if a product name or price changes, embedded copies must be updated), and document size growth (subject to MongoDB's 16MB document size limit).

Referencing (Normalization):

Pros: Eliminates data redundancy, guarantees strong data consistency (updates happen in one central document), and prevents document size bloat for unbounded relationships (like unlimited historical purchases).

Cons: Requires multiple network round-trips or $lookup pipeline stages to assemble full objects, resulting in higher query latency for UI rendering.

3. **How does this schema handle products with different attributes?**

This schema leverages MongoDB's polymorphic sub-document pattern through the flexible attributes object.

Because NoSQL databases are schema-agnostic, products across distinct industries do not need to share a rigid set of columns:

An Aerospace product can store structural parameters like tensile_strength_mpa, weight_grams, and 3D dimensions (x, y, z).

A Medical product can dynamically store domain-specific flags such as sterilizable: true, layer_height_microns, or color.

This flexibility allows ACME to introduce entirely new product types with unique manufacturing specs into the products collection without requiring schema migrations, ALTER TABLE statements, or sparse NULL columns.

## Recommendation Queries

### Customers who bought X also bought Y (Required)

```python
# TODO: write your "people also purchased" recommendation query (create more cells as needed)
# =========================================================================
# Method 1: COLLABORATIVE FILTERING VIA ALSO_BOUGHT EDGES
# =========================================================================


also_bought_query = """
MATCH (c:Customer {customer_id: $customer_id}) - [:PURCHASED] -> (p1:Product) - [r:ALSO_BOUGHT] -> (rec:Product)
WHERE NOT (c) - [:PURCHASED] -> (rec)
RETURN p1.product_name AS BecauseYouBought,
       rec.product_name AS RecommendedProduct,
       r.confidence AS ConfidenceScore,
       r.support_count AS SupportCount
ORDER BY ConfidenceScore DESC
"""

try:
    print("=" *80)
    print("METHOD 1:Collaborative filtering via precomputed ALSO_BOUGHT edges")
    print("=" * 80)
    results = neo4j_conn.execute_query(also_bought_query, {"customer_id": 2})
    
    for r in results:
        print(f"  Since you bought '{r['BecauseYouBought']}':")
        print(f"    ↳ We recommend '{r['RecommendedProduct']}'")
        print(f"      [Confidence: {r['ConfidenceScore']:.0%}, Co-purchased {r['SupportCount']} times]\n")
except Exception as e:
    print(f"Error executing Method 1: {e}")
    
    
# =========================================================================
# METHOD 2 (OPTIMISED): DYNAMIC COLLABORATIVE FILTERING
# =========================================================================

dynamic_collab_query = """
// Step 1: Single-path traversal tracing: My Purchases ➔ Shared Buyers ➔ Their Purchases
MATCH (c1:Customer {customer_id: $customer_id})-[:PURCHASED]->(p1:Product)<-[:PURCHASED]-(c2:Customer)-[:PURCHASED]->(p2:Product)

// Step 2: Ensure we exclude Alice herself, self-matching items, and items she already bought
WHERE c1 <> c2 
  AND p1 <> p2 
  AND NOT (c1)-[:PURCHASED]->(p2)

// Step 3: Aggregate the overlapping customer count and sort by recommendation strength
RETURN p1.product_name AS BecauseYouBought,
       p2.product_name AS RecommendedProduct,
       count(DISTINCT c2) AS SharedBuyersCount
ORDER BY SharedBuyersCount DESC
"""

try:
    print("=" * 80)
    print("METHOD 2 (OPTIMISED): DYNAMIC CO-PURCHASE TRAVERSAL")
    print("=" * 80)
    results = neo4j_conn.execute_query(dynamic_collab_query, {"customer_id": 2})
    
    if results:
        for r in results:
            print(f"  Since you bought '{r['BecauseYouBought']}':")
            print(f"    ↳ We recommend '{r['RecommendedProduct']}'")
            print(f"      [Recommended by {r['SharedBuyersCount']} other customer(s) who bought both]\n")
    else:
        print("  No dynamic co-purchases found (requires overlapping customer transactions).")
    print("=" * 80)
except Exception as e:
    print(f"Error executing Method 2: {e}")
```

## Graph Design Decisions

1. **What nodes and relationships did you create and why?**

Node Types:

Customer: Represents buyers in the system (properties: customer_id, first_name, last_name, email, phone).

Product: Represents the 3D-printed catalog items (properties: product_id, product_name, material, tensile_strength_mpa, sterilizable, infill_density).

Industry: Represents industry sectors served by ACME (properties: name, e.g., Aerospace, Medical)

Relationship Types:

(:Customer)-[:WORKS_IN]->(:Industry): Connects customers to their professional industry domain for domain-targeted filtering and recommendations.

(:Product)-[:SERVES]->(:Industry): Categorizes products by the specific industry application they are engineered for.

(:Customer)-[:PURCHASED]->(:Product): Tracks transactional history between customers and products, capturing point-in-time purchase metadata.

(:Product)-[:ALSO_BOUGHT]->(:Product): Direct product-to-product edge used for collaborative filtering recommendations.

Why: Graph databases excel at traversing relationships. Separating entities into discrete node labels connected by semantic relationships allows us to perform real-time recommendation traversals without heavy relational JOIN operations.

2. **How does your graph structure enable the "people also purchased" recommendation?**

Our graph structure enables collaborative filtering through two distinct patterns:Pre-computed Edges (ALSO_BOUGHT):Traverses directly from a product the customer bought to candidate items via precomputed co-purchase links:$$\text{(Customer)} \xrightarrow{\text{:PURCHASED}} \text{(Product A)} \xrightarrow{\text{:ALSO\_BOUGHT}} \text{(Product B)}$$This approach provides ultra-low latency reads ($O(1)$ lookup per item) for real-time recommendation engines.Dynamic Path Traversal:Traverses through shared buyer history in real time:$$\text{(Customer 1)} \xrightarrow{\text{:PURCHASED}} \text{(Product A)} \xleftarrow{\text{:PURCHASED}} \text{(Customer 2)} \xrightarrow{\text{:PURCHASED}} \text{(Product B)}$$By finding overlapping buyers ($\text{Customer 2}$) who bought the same initial product ($\text{Product A}$), the graph dynamically identifies and ranks unpurchased peer items ($\text{Product B}$)

3. **What properties did you add to relationships and why?**

PURCHASED Relationship Properties:

purchase_id, date, quantity, unit_price, total_price, status.

Why: In graph modeling, edge-level attributes capture the context of an interaction. Storing transaction details directly on the PURCHASED edge avoids creating separate, heavy transaction nodes for simpler models while allowing temporal filtering (e.g., filtering purchases within a specific date range).

ALSO_BOUGHT Relationship Properties:

confidence (float, e.g., 0.85), support_count (integer, e.g., 142).

Why: Storing statistical weights directly on the relationship allows the Cypher engine to ORDER BY ConfidenceScore DESC and restrict recommendations to high-confidence co-purchases without re-calculating statistics on every query.

4. **How would you handle graph growth as more customers and products are added?**

Indexing & Constraints:

Create uniqueness constraints and B-tree indexes on primary identifiers (Customer.customer_id, Product.product_id, Industry.name) to guarantee fast MATCH lookups as node counts scale into millions.

Relationship Pruning & Thresholding:

Filter out low-confidence ALSO_BOUGHT edges (e.g., only retain edges where confidence >= 0.5 or support_count >= 10) during offline batch ETL runs to prevent excessive graph dense-node degradation (supernodes).

Graph Partitioning & Read Replicas:

Deploy a Neo4j Causal Clustering topology with dedicated Read Replicas to scale read-heavy recommendation traffic horizontally across servers.
